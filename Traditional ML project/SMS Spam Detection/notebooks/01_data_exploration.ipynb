```python
import pandas as pd 
import matplotlib.pyplot as plt
data = pd.read_csv("SMSSpamCollection", sep="\t", header=None, names=["label", "message"])


label_count = data["label"].value_counts()

message_length = data['message'].str.len()
data['message_length'] = data['message'].str.len()

data.head()

average_length = data.groupby('label')['message_length'].mean()

plt.figure(figsize=(8, 5))
plt.bar(average_length.index, average_length.values)

plt.title('Length of Messages')
plt.xlabel("Label")
plt.ylabel('length')

plt.show()
```
